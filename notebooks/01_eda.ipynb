## Objetivo

Realizar la carga y exploración inicial del dataset Wall-Following Robot Navigation,
identificando su estructura, tipos de datos, variable objetivo y distribución de clases.

```python
from pathlib import Path
import pandas as pd
```

```python
project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

data_path = project_root / "data" / "raw" / "sensor_readings_24.data"

print("Raíz del proyecto:", project_root)
print("Ruta del dataset:", data_path)
print("¿Existe el dataset?:", data_path.exists())
```

```python
sensor_columns = [f"US{i}" for i in range(1, 25)]
columns = sensor_columns + ["Class"]

print(columns)
print("Número de columnas:", len(columns))
```

```python
df = pd.read_csv(
    data_path,
    header=None,
    names=columns
)

df.head()
```

## Distribución de la variable objetivo

Se analiza cuántas observaciones pertenecen a cada clase de navegación

```python
df["Class"].value_counts()
```

## Valores faltantes

Se verifica si existen datos ausentes en alguna de las variables.
Los valores faltantes deben identificarse antes de entrenar modelos,
ya que pueden requerir eliminación o imputación.

```python
missing_values = df.isnull().sum()

print("Valores faltantes por columna:")
print(missing_values)

print("\nTotal de valores faltantes:", missing_values.sum())
```

## Registros duplicados

Se comprueba la existencia de filas completamente repetidas.
Los duplicados pueden influir en el análisis y deben identificarse
antes de decidir si es necesario eliminarlos.

```python
duplicate_rows = df.duplicated().sum()

print("Número de filas duplicadas:", duplicate_rows)
```

## Balance porcentual de clases

Además del número de observaciones por clase, se calcula su proporción
porcentual para evaluar con mayor claridad el nivel de desbalance
de la variable objetivo.

```python
class_counts = df["Class"].value_counts()
class_percentages = df["Class"].value_counts(normalize=True) * 100

class_balance = pd.DataFrame({
    "Cantidad": class_counts,
    "Porcentaje (%)": class_percentages.round(2)
})

class_balance
```

### Interpretación del balance de clases

La distribución de la variable objetivo no es uniforme. Las clases `Move-Forward`
(40.41 %) y `Sharp-Right-Turn` (38.43 %) concentran la mayor parte de las
observaciones, mientras que `Slight-Right-Turn` representa el 15.14 % y
`Slight-Left-Turn` únicamente el 6.01 %.

Por lo tanto, el dataset presenta desbalance de clases. Este aspecto deberá
considerarse posteriormente durante la división de los datos y la evaluación
de los modelos, especialmente para evitar que un buen rendimiento global
oculte un desempeño deficiente en las clases menos representadas.

## Conclusiones de la exploración inicial

El dataset contiene **5.456 observaciones** y **25 columnas**, de las cuales
24 corresponden a lecturas de sensores ultrasónicos (`US1` a `US24`) y una

Las variables de los sensores son numéricas y la variable objetivo es

La variable objetivo presenta cuatro clases de navegación:

La distribución de clases no es uniforme, por lo que existe desbalance,
