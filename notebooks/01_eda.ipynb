# EDA completo

En esta sección se analiza visualmente la distribución de las variables,
la presencia de posibles valores atípicos, las relaciones entre los sensores
y el comportamiento de la variable objetivo.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

## Distribución de sensores seleccionados

Se visualiza la distribución de seis sensores ultrasónicos distribuidos
a lo largo del conjunto de 24 sensores, con el objetivo de observar
la concentración de las mediciones, su rango y posibles asimetrías.

```python
selected_sensors = ["US1", "US5", "US9", "US13", "US17", "US21"]

df[selected_sensors].hist(
    bins=30,
    figsize=(12, 8)
)

plt.suptitle("Distribución de sensores ultrasónicos seleccionados")
plt.tight_layout()
plt.show()
```

### Interpretación de las distribuciones

Los sensores analizados presentan distribuciones no simétricas y comportamientos
diferentes entre sí. Varias variables concentran una gran cantidad de observaciones
en distancias bajas, mientras que otras presentan múltiples grupos de valores.

En los sensores `US5`, `US9` y `US13` se observa además una acumulación importante
de mediciones alrededor de `5.0`. Este comportamiento también aparece, aunque con
menor frecuencia, en otros sensores.

La presencia repetida de este valor sugiere que podría corresponder a un límite
máximo de medición del sensor, por lo que no debe considerarse automáticamente
como un valor atípico. Este aspecto deberá verificarse antes del preprocesamiento.

En general, las variables no presentan una distribución normal y varias muestran
asimetría positiva y diferentes concentraciones de mediciones.

## Detección visual de posibles valores atípicos

Se utilizan diagramas de caja para analizar la dispersión de cuatro sensores
y detectar visualmente observaciones que se encuentren alejadas del rango central
de sus mediciones.

Los puntos identificados por un boxplot como posibles valores atípicos no se
eliminarán automáticamente, ya que primero debe determinarse si corresponden
a errores o a mediciones reales del comportamiento del robot.

```python
boxplot_sensors = ["US1", "US5", "US13", "US17"]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, sensor in zip(axes.flatten(), boxplot_sensors):
    sns.boxplot(
        y=df[sensor],
        ax=ax
    )

    ax.set_title(f"Boxplot de {sensor}")
    ax.set_ylabel("Distancia")

plt.tight_layout()
plt.show()
```

### Interpretación de los boxplots

Los sensores presentan comportamientos diferentes en cuanto a dispersión y
posibles valores atípicos.

`US5` muestra una distribución amplia sin observaciones claramente separadas
del rango principal. Por otra parte, `US1` y `US13` presentan algunas mediciones
altas que son identificadas por el boxplot como posibles valores atípicos.

El sensor `US17` muestra el comportamiento más asimétrico: la mayoría de sus
mediciones se concentra en distancias bajas, mientras que existe un número
considerable de observaciones superiores alejadas del rango central.

Estos valores no se eliminarán automáticamente, ya que pueden representar
situaciones reales de navegación y no necesariamente errores de medición.

## Correlación entre sensores ultrasónicos

Se analiza la relación lineal entre las 24 variables de entrada mediante una
matriz de correlación. Esto permite identificar sensores cuyas mediciones tienden
a variar de forma similar o en direcciones opuestas.

```python
sensor_data = df[sensor_columns]

correlation_matrix = sensor_data.corr()

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    square=True,
    linewidths=0.3
)

plt.title("Matriz de correlación entre sensores ultrasónicos")
plt.tight_layout()
plt.show()
```

### Interpretación de la matriz de correlación

La matriz muestra que existen grupos de sensores con correlaciones positivas,
especialmente entre sensores cercanos entre sí. Esto indica que varias mediciones
tienden a cambiar de manera similar cuando observan regiones relacionadas del entorno.

También se observan correlaciones negativas entre determinados grupos de sensores,
lo que sugiere comportamientos opuestos según la orientación del robot y la posición
de los obstáculos o paredes.

Por lo tanto, las variables de entrada no son completamente independientes.
Estas relaciones podrán ser relevantes posteriormente durante el análisis de
características y la reducción dimensional.

## Correlaciones más fuertes entre sensores

Se identifican numéricamente las parejas de sensores con mayor correlación
positiva y negativa para complementar la interpretación visual del heatmap.

```python
upper_triangle = np.triu(
    np.ones(correlation_matrix.shape, dtype=bool),
    k=1
)

corr_pairs = (
    correlation_matrix
    .where(upper_triangle)
    .stack()
    .dropna()
    .sort_values()
)

print("Correlaciones negativas más fuertes:")
print(corr_pairs.head(5))

print("\nCorrelaciones positivas más fuertes:")
print(corr_pairs.tail(5).sort_values(ascending=False))
```

### Interpretación de las correlaciones más fuertes

Las correlaciones positivas más elevadas se presentan entre `US4-US5`,
`US9-US10`, `US8-US10`, `US8-US9` y `US17-US18`, con valores comprendidos
aproximadamente entre 0.51 y 0.63.

Esto indica que algunos sensores presentan comportamientos lineales similares,
aunque ninguna de las correlaciones observadas es cercana a 1, por lo que no
existe una relación lineal perfecta entre las variables analizadas.

Las correlaciones negativas más fuertes presentan magnitudes menores, entre
aproximadamente -0.30 y -0.37. Esto indica relaciones inversas moderadas entre
determinados sensores.

En conjunto, los resultados muestran que existen relaciones entre varias
variables de entrada, pero no una redundancia lineal extrema. Estas relaciones
podrán ser relevantes posteriormente en técnicas como reducción dimensional.

## Distribución visual de las clases de navegación

Se representa gráficamente la cantidad de observaciones pertenecientes a cada
acción del robot para visualizar el desbalance existente en la variable objetivo.

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="Class",
    order=df["Class"].value_counts().index
)

plt.title("Distribución de las clases de navegación")
plt.xlabel("Clase")
plt.ylabel("Número de observaciones")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

### Interpretación de la distribución de clases

La representación gráfica confirma que la variable objetivo presenta una
distribución desbalanceada.

Las clases `Move-Forward` y `Sharp-Right-Turn` concentran la mayor cantidad
de observaciones, mientras que `Slight-Right-Turn` y especialmente
`Slight-Left-Turn` poseen una representación considerablemente menor.

Este desbalance deberá considerarse durante la división del dataset y la
evaluación de los modelos para evitar que las clases mayoritarias dominen
las métricas globales de rendimiento.

## Generación del resumen del EDA

Se genera un archivo JSON con los principales resultados del análisis
exploratorio. Este archivo permitirá conservar de forma estructurada
información sobre las dimensiones del dataset, valores faltantes,
estadísticas descriptivas y distribución de clases.

```python
import json

eda_summary = {
    "dataset": {
        "instances": int(df.shape[0]),
        "attributes": int(df.shape[1]),
        "input_features": 24,
        "target": "Class"
    },
    "missing_values": {
        column: int(value)
        for column, value in df.isnull().sum().items()
    },
    "descriptive_statistics": json.loads(
        df[sensor_columns].describe().to_json()
    ),
    "class_distribution": {
        class_name: int(count)
        for class_name, count in df["Class"].value_counts().items()
    }
}

summary_path = project_root / "data" / "processed" / "eda_summary.json"

with open(summary_path, "w", encoding="utf-8") as file:
    json.dump(
        eda_summary,
        file,
        indent=4,
        ensure_ascii=False
    )

print("Archivo generado en:")
print(summary_path)
print("\n¿Existe?:", summary_path.exists())
```
